```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
import streamlit as st
```

```python
# loading the diabetes dataset to a pandas DataFrame
dataset = pd.read_csv('diabetes.csv') 
```

```python

dataset.head()
#shows the first 5 rows
```

```python
# number of rows and Columns in this dataset
st.text(dataset.shape)
```

```python
# getting the statistical measures of the data
dataset.describe()
```

```python
dataset['Outcome'].value_counts()
```

```python
dataset.groupby('Outcome').mean()
# to get the mean value for each feature for both pesrons with diabetes 
#and without
```

```python
#seperating the data and the labels
X = dataset.drop(columns = 'Outcome', axis=1) #axis = 1 for droping a column // axis = 0 if for a row
Y = dataset['Outcome'] #label
```

```python
#Standerization for better performance
scaler = StandardScaler()
```

```python
scaler.fit(X)
```

```python
standardized_data = scaler.transform(X)
```

```python
print(standardized_data)
```

```python
X = standardized_data
Y = dataset['Outcome']
```

```python
print(Y)
print(X)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
#Support vector machine
classifier = svm.SVC(kernel='linear')
```

```python
#training the support vector Machine Classifier
classifier.fit(X_train, Y_train)
```

```python
# accuracy score on the training data
X_train_prediction_smv = classifier.predict(X_train)
training_data_accuracy_smv = accuracy_score(X_train_prediction_smv, Y_train)

# accuracy score on the test data
X_test_prediction_smv = classifier.predict(X_test)
test_data_accuracy_smv = accuracy_score(X_test_prediction_smv, Y_test)
```

```python
print('Accuracy score of the training data : ', training_data_accuracy_smv)
print('Accuracy score of the test data : ', test_data_accuracy_smv)
```

```python
#Naive Bayes Model

nb = GaussianNB()
nb.fit(X_train,Y_train)
```

```python
# accuracy score on the training data
X_train_prediction_naive = nb.predict(X_train)
training_data_accuracy_naive = accuracy_score(X_train_prediction_naive, Y_train)

# accuracy score on the test data
X_test_prediction_naive = nb.predict(X_test)
test_data_accuracy_naive = accuracy_score(X_test_prediction_naive, Y_test)
```

```python
print('Accuracy score of the training data : ', training_data_accuracy_naive)
print('Accuracy score of the test data : ', test_data_accuracy_naive)
```

```python
#DecisionTree model

dt = DecisionTreeClassifier(max_depth = 6)
dt.fit(X_train,Y_train)
```

```python
# accuracy score on the training data
X_train_prediction_dt = dt.predict(X_train)
training_data_accuracy_dt = accuracy_score(X_train_prediction_dt, Y_train)

# accuracy score on the test data
X_test_prediction_dt = dt.predict(X_test)
test_data_accuracy_dt = accuracy_score(X_test_prediction_dt, Y_test)
```

```python
print('Accuracy score of the training data : ', training_data_accuracy_dt)
print('Accuracy score of the test data : ', test_data_accuracy_dt)
```
